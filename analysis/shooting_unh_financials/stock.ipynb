```python
news_broke_start_time = pd.Timestamp("2024-12-04 07:36:00", tz="US/Eastern")
news_broke_end_time = pd.Timestamp("2024-12-04 09:00:00", tz="US/Eastern")
news_broke_label = "News of shooting breaking"
```

```python
import altair as alt

price_min = df["average_price"].min()
price_max = df["average_price"].max()
price_buffer = (price_max - price_min) * 0.05

line = (
    alt.Chart(df)
    .mark_line()
    .encode(
        x=alt.X(
            "ts_minute:T",
            title="Minute block (US Eastern)",
            axis=alt.Axis(
                format="%b %-d, %-I:%M%p",
                labelAngle=45,
            ),
        ),
        y=alt.Y(
            "average_price:Q",
            title="Average Price (USD)",
            axis=alt.Axis(format="$,f"),
            scale=alt.Scale(
                domain=[price_min - price_buffer, price_max + price_buffer],
                nice=False,
            ),
        ),
        tooltip=[
            alt.Tooltip("ts_minute:T", title="Time", format="%Y-%m-%d %H:%M %Z"),
            alt.Tooltip("average_price:Q", title="Average Price", format="$,.2f"),
        ],
    )
)

event_data = pd.DataFrame(vlines)
events = (
    alt.Chart(event_data)
    .mark_rule(
        color="crimson",
        strokeDash=[6, 4],
        strokeWidth=2,
    )
    .encode(
        x="time:T",
    )
)

event_labels = (
    alt.Chart(event_data)
    .mark_text(
        align="right",
        baseline="top",
        color="crimson",
        dx=-5,
        dy=5,
    )
    .encode(
        x="time:T",
        y=alt.value(0),
        text="label:N",
    )
)

news_window = (
    alt.Chart(
        pd.DataFrame(
            {
                "start": [news_broke_start_time],
                "end": [news_broke_end_time],
                "label": [news_broke_label],
            }
        )
    )
    .mark_rect(
        color="crimson",
        opacity=0.25,
    )
    .encode(
        x="start:T",
        x2="end:T",
    )
)

news_label = (
    alt.Chart(
        pd.DataFrame(
            {
                "time": [news_broke_start_time],
                "label": [news_broke_label],
            }
        )
    )
    .mark_text(
        align="left",
        baseline="top",
        color="crimson",
        dx=5,
        dy=5,
    )
    .encode(
        x="time:T",
        y=alt.value(0),
        text="label:N",
    )
)

charts = line + news_window + events + event_labels + news_label

chart = charts.properties(
    title="UNH Stock Price Over Time",
    width="container",
    height=450,
).interactive()

chart
```
