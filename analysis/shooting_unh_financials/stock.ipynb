```python
import plotly.express as px

fig = px.line(
    df,
    x="ts_minute",
    y="average_price",
    title="UNH Stock Price Over Time",
    labels={
        "ts_minute": "Minute block (US Eastern)",
        "average_price": "Average Price (USD)",
    },
)

fig.update_traces(mode="lines")

for line in vlines:
    fig.add_vline(
        x=line["time"],
        line_dash="dash",
        line_color="crimson",
        annotation_text=line["label"],
        annotation_position="bottom left",
        annotation_font_color="crimson",
    )

fig.add_vrect(
    x0=news_broke_start_time,
    x1=news_broke_end_time,
    annotation_text=news_broke_label,
    annotation_position="top left",
    fillcolor="crimson",
    opacity=0.25,
    line_width=0,
)

fig.show()
```
